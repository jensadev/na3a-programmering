```python
print("Välkommen till additionsräknaren!")

# Fråga användaren om det första talet
num1_str = input("Ange det första talet: ")

# Konvertera strängen till ett heltal
num1 = int(num1_str)

# Fråga användaren om det andra talet
num2_str = input("Ange det andra talet: ")

# Konvertera strängen till ett heltal
num2 = int(num2_str)

# Utför additionen
summa = num1 + num2

# Skriv ut resultatet
print(f"Summan av {num1} och {num2} är: {summa}")
```
